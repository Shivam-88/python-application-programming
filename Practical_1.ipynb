```python
float_num = 5.9
int_from_float = int(float_num)
print("Float to Integer:", int_from_float)

string_num = "42"
int_from_string = int(string_num)
print("String to Integer:", int_from_string)

bool_val = True
int_from_bool = int(bool_val)
print("Boolean to Integer:", int_from_bool)
```
