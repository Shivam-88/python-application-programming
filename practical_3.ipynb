```python
inventory = {}
```

```python
p1_name = "Mobile Phone"
p1_quantity = 5
p1_price = 20000
p1_year = 2020
```

```python
inventory["ProductNo1"] = {
    "Product Name": p1_name,
    "Product Quantity": p1_quantity,
    "Product Price": p1_price,
    "Product Release Year": p1_year,
}
```

```python
p2_name = "Laptop"
p2_quantity = 3
p2_price = 50000
p2_year = 2021
```

```python
inventory["ProductNo2"] = {
    "Product Name": p2_name,
    "Product Quantity": p2_quantity,
    "Product Price": p2_price,
    "Product Release Year": p2_year,
}
```

```python
print(inventory)
```

```python
has_p1_year = "Product Release Year" in inventory["ProductNo1"]
has_p2_year = "Product Release Year" in inventory["ProductNo2"]
print(f"Product 1 Year Exists: {has_p1_year}")
print(f"Product 2 Year Exists: {has_p2_year}")
```

```python
del inventory["ProductNo1"]["Product Release Year"]
del inventory["ProductNo2"]["Product Release Year"]
print(inventory)
```

```python
prices = (250, 300, 150, 400, 100, 350, 200)
```

```python
highest = max(prices)
lowest = min(prices)
print(f"Highest: {highest}")
print(f"Lowest: {lowest}")
```

```python
total = sum(prices)
print(f"Total Sum: {total}")
```

```python
sorted_prices = sorted(prices)
print(sorted_prices)
```

```python
try:
    prices[0] = 500
except TypeError as error:
    print(f"Error: {error}")
    print("Why: Tuples are immutable and cannot be changed.")
```
