```python
def zipper_file(filename, strings):

    with open(filename, "w") as file:

        for i in range(0, len(strings), 2):

            line = strings[i] + strings[i + 1]

            file.write(line + "\n")

strings = ["Hello", "World", "Python", "Programming", "Good", "Morning"]

zipper_file("output.txt", strings)

print("File created successfully!")
```

```python
def reformat_date(dates):

    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    result = []

    for date in dates:

        month, day, year = date.split("/")

        sentence = "Office Hours on " + months[month] + " " + day + ", " + year

        result.append(sentence)

    return result

dates = ["01/15/2025", "03/22/2025", "12/05/2025"]

answer = reformat_date(dates)

for item in answer:
    print(item)
```
