```python
def safe_divide(numerator, denominator):
    try:
        result = numerator / denominator
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by Zero.")
        return None
try:
    user_numerator = float(input("Enter the numerator: "))
    user_denominator = float(input("Enter the denominator: "))
    output = safe_divide(user_numerator, user_denominator)
    if output is not None:
        print(f"Result: {output}")
except ValueError:
    print("Invalid input! Please enter numbers only.")


import math

def calculate_square_root(number1):
    try:
        result = math.sqrt(number1)
        return result
    except ValueError:
        print("Invalid input! Please enter a positive integer or a float value.")
        return None

def perform_complex_task(num):
    try:
        result = num / (num - 5)
        return result
    except Exception as e:
        print("An error occurred during calculation.")
        return None
```
