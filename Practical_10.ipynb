```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset("iris")

print(iris.head())

print(iris.info())
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=iris,
    x="sepal_length",
    y="sepal_width",
    hue="species"
)
plt.title("Iris Scatter Plot")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=iris,
    x="petal_length",
    hue="species",
    fill=True,
    alpha=0.3
)
plt.title("Iris Density Plot")
plt.xlabel("Petal Length")
plt.ylabel("Density")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=iris,
    x="species",
    y="petal_length"
)

plt.title("Iris Violin Plot")
plt.xlabel("Species")
plt.ylabel("Petal Length")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = iris.select_dtypes(include="number").corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Iris Correlation Heatmap")

plt.show()
```

```python
sns.jointplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    height=7
)

plt.suptitle("Iris Joint Plot", y=1.02)

plt.show()
```

```python
sns.pairplot(
    iris,
    hue="species",
    diag_kind="hist"
)

plt.show()
```
